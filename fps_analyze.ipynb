FPS plots for all experiement runs in a folder

```python
import json
import os
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np


def plot_experiment_round(experiment_dir_path):
    """Navigates an experiment round folder structure, creates a grid layout

    where plots stack 5 vertically per column, moving left-to-right.

    Structure:
    experimentRound1/
      ├── test1/  -> Column 1, Row 1
      ├── test2/  -> Column 1, Row 2
      ...
      ├── test5/  -> Column 1, Row 5
      ├── test6/  -> Column 2, Row 1
      └── ...
    """
    base_dir = Path(experiment_dir_path)

    # Helper function to auto-wrap long text titles cleanly
    def wrap_text(text, max_chars=22):
        if len(text) <= max_chars:
            return text
        words = text.replace("_", " ").replace("-", " ").split(" ")
        lines = []
        current_line = []
        current_len = 0

        for word in words:
            if current_len + len(word) + 1 > max_chars:
                if current_line:
                    lines.append(" ".join(current_line))
                current_line = [word]
                current_len = len(word)
            else:
                current_line.append(word)
                current_len += len(word) + 1
        if current_line:
            lines.append(" ".join(current_line))
        return "\n".join(lines)

    # 1. Find and sort all "test" subdirectories containing a "tracking_results" folder
    test_dirs = sorted(
        [
            d
            for d in base_dir.iterdir()
            if d.is_dir() and (d / "tracking_results").is_dir()
        ]
    )

    if not test_dirs:
        print(f"No valid test directories found in {base_dir}")
        return

    num_plots = len(test_dirs)

    # 2. Configure column-major grid layout: Always max 5 rows, columns grow dynamically
    nrows = min(5, num_plots)
    ncols = (num_plots + nrows - 1) // nrows  # Integer ceiling division

    # Sizing adjustments: Widened columns to handle 5-vertical stack layout nicely
    fig, axes = plt.subplots(
        nrows=nrows, ncols=ncols, figsize=(6 * ncols, 5 * nrows), squeeze=False
    )

    # CRITICAL: Flatten using 'F' (Fortran-order) to iterate column-by-column
    flat_axes = axes.flatten(order="F")

    # 3. Iterate through each test folder and its corresponding subplot axis
    for i, test_dir in enumerate(test_dirs):
        ax = flat_axes[i]
        tracking_dir = test_dir / "tracking_results"

        json_files = sorted(tracking_dir.glob("*.json"))

        # Wrap long names and apply title
        wrapped_title = wrap_text(f"Test: {test_dir.name}", max_chars=22)
        ax.set_title(wrapped_title, fontsize=10, weight="bold", pad=10)

        if not json_files:
            ax.text(
                0.5,
                0.5,
                "No JSON files found",
                ha="center",
                va="center",
                transform=ax.transAxes,
            )
            continue

        # 4. Plot each JSON file inside this specific test subplot
        for json_path in json_files:
            try:
                with open(json_path, "r") as f:
                    data = json.load(f)

                frames = data.get("frames", [])
                if not frames:
                    continue

                timestamps = [f["packet_timestamp_utc_ms"] for f in frames]
                deltas = np.diff(timestamps)
                fps_values = [1000.0 / d if d > 0 else 0 for d in deltas]

                label_name = json_path.stem

                ax.plot(
                    range(len(fps_values)),
                    fps_values,
                    label=f"{label_name} (Avg: {np.mean(fps_values):.1f})",
                )

            except Exception as e:
                print(f"Error reading {json_path}: {e}")

        # Subplot styling
        ax.set_ylabel("FPS")
        ax.set_xlabel("Frame Index")
        ax.grid(True)

        # Legend bottom placement
        ax.legend(
            loc="upper center",
            bbox_to_anchor=(0.5, -0.15),
            ncol=1,
            fontsize=8,
            frameon=True,
        )

    # 5. Hide any unused subplot axes at the bottom-right of the grid
    for j in range(num_plots, len(flat_axes)):
        fig.delaxes(flat_axes[j])

    # Clean up layout adjustments and save
    plt.tight_layout()
    plt.savefig(f"./{base_dir.name}/{base_dir.name}_fps_plot.png", dpi=300)
    plt.show()


# --- Usage ---
plot_experiment_round('eval2')
```
